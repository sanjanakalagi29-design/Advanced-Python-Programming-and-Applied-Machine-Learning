```python
def Student(*marks):
  sum=0
  avg=len(marks)
  for i in marks:
   sum=sum+i
  print("Total sum value is :",sum)
  print("Mean value is :",sum/avg)
  print("Maximum value is :",max(marks))
  print("Minimum value is :",min(marks))
Student(1,2,3,4,5)
Student(5,10,25,40,55,70,95,110)
```
