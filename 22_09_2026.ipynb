```python
from typing import Literal
#1)import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score

# 2)create DataSets
data={'House (sqrt)':[50,100,150,200,250,300,350],
      'Price (Lakshs)':[20,32,40,50,65,80,95]
      }

df=pd.DataFrame(data)
print(type(data))
df.dtypes
#df.info()

# 3)find out input and output variables
X = df[['House (sqrt)']]
Y = df['Price (Lakshs)']

# 4)Train - test model
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size = 0.3)

# 5)object creation of algorithms
model=LinearRegression()

# 6) call fit() function of input and output variable
model.fit(X_train,Y_train)

# 7) predict New value
print("Prediction of new value is :",model.predict([[1400]]))

# Calculate Metrics
MSE=mean_squared_error(Y_test,X_test)
print("Mean_squared_error value is",mean_squared_error(Y_test,X_test))
print("R2_score value is:",r2_score(Y_test,X_test))

pridiction=model.predict(X_train)
# 9) Data Visualizationa(Using Matplotlib)
plt.figure(figsize=(10,5))
plt.scatter(X,Y,color="#AB03A9",marker="X",edgecolor="pink",linewidths=0.5,alpha=0.7,label="Data Points")
plt.plot(X,model.predict(X),label="Regression Line",color="#3E0F8D",linestyle='--')
plt.title("Linear Regression")
plt.xlabel("House (sqrt)")
plt.ylabel("Price (Lakshs)")
plt.grid(linestyle="--")
plt.legend()
plt.show()

```
